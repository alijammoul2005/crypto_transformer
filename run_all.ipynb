## Cell 7: Interactive Decryption (Optional)

Try decrypting your own ciphertexts!

```python
# Interactive decryption
import sys
sys.path.append('/kaggle/working/crypto_transformer')

from inference.decrypt import load_model, decrypt
from data.generate_data import generate_key, encrypt

# Load model
load_model()

# Example: Create your own encrypted text
plaintext = "the transformer learns patterns"
key = generate_key()
ciphertext = encrypt(plaintext, key)

print(f"Original:   {plaintext}")
print(f"Encrypted:  {ciphertext}")

# Decrypt
decrypted = decrypt(ciphertext, method='greedy')
print(f"Decrypted:  {decrypted}")

# Try with your own text
print("\n" + "="*80)
print("Try your own:")
print("="*80)

custom_plaintext = "deep learning is amazing"  # Change this
custom_key = generate_key()
custom_cipher = encrypt(custom_plaintext, custom_key)

print(f"Your plaintext:  {custom_plaintext}")
print(f"Encrypted:       {custom_cipher}")

custom_decrypted = decrypt(custom_cipher, method='greedy')
print(f"Decrypted:       {custom_decrypted}")

# Calculate accuracy
from evaluation.metrics import character_accuracy
acc = character_accuracy(custom_decrypted, custom_plaintext)
print(f"Accuracy:        {acc:.2f}%")
```
