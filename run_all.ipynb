# Transformer-Based Substitution Cipher Cryptanalysis

## **NEW KEY PREDICTION APPROACH** (2020 Neural Cryptanalysis Paper)

This notebook demonstrates an improved deep learning system for breaking substitution ciphers using **key prediction** instead of direct plaintext prediction.

## Overview

- **Task**: Predict the 26-letter substitution KEY, then apply it deterministically
- **Architecture**: Encoder-Decoder Transformer
- **Dataset**: WikiText-103 with fresh random keys per epoch
- **Training**: 300k samples, fresh keys generated on-the-fly
- **Evaluation**: Key accuracy, perfect key recovery rate, character accuracy

## Why This Approach is Better

1. **Fixed output length**: 26 letters vs 200 letters → much easier task
2. **Deterministic decryption**: Once key is predicted, apply it perfectly
3. **Easier to evaluate**: Exact key matching + decryption quality
4. **Faster training**: Smaller output space, faster convergence
5. **Fresh keys**: Maximum diversity across epochs

1. Clone code from GitHub

## Cell 8: Interactive Decryption (Optional)

Try decrypting your own ciphertexts!

```python
# Interactive decryption
import sys
import os

# Detect environment
try:
    import google.colab
    base_dir = '/content'
except:
    base_dir = os.getcwd()

# Add to path
sys.path.insert(0, os.path.join(base_dir, 'crypto_transformer'))

from inference.decrypt import load_model, decrypt
from data.generate_data import generate_key, encrypt, key_to_string

print("="*80)
print("Loading model for interactive demo...")
print("="*80)

# Load model
load_model()

# Example: Create your own encrypted text
plaintext = "the transformer predicts substitution keys" + ' ' * 158
key_dict = generate_key()
true_key = key_to_string(key_dict)
ciphertext = encrypt(plaintext, key_dict)

print("\n" + "="*80)
print("Example 1: Random Encryption")
print("="*80)
print(f"Original:   {plaintext.strip()}")
print(f"Encrypted:  {ciphertext[:80]}...")
print(f"True key:   {true_key}")

# Decrypt using model
result = decrypt(ciphertext, method='greedy')
predicted_key = result['key']
decrypted = result['plaintext']

print(f"\nPredicted key: {predicted_key}")
print(f"Decrypted:     {decrypted.strip()}")

# Calculate accuracy
from evaluation.metrics import key_accuracy, character_accuracy
key_acc = key_accuracy(predicted_key, true_key)
char_acc = character_accuracy(decrypted.strip(), plaintext.strip())
print(f"\nKey accuracy:       {key_acc:.2f}%")
print(f"Character accuracy: {char_acc:.2f}%")

# Try with your own text
print("\n" + "="*80)
print("Example 2: Custom Text")
print("="*80)

custom_plaintext = "deep learning solves cryptography" + ' ' * 166  # Pad to 200 chars
custom_key_dict = generate_key()
custom_true_key = key_to_string(custom_key_dict)
custom_cipher = encrypt(custom_plaintext, custom_key_dict)

print(f"Your plaintext: {custom_plaintext.strip()}")
print(f"Encrypted:      {custom_cipher[:80]}...")
print(f"True key:       {custom_true_key}")

custom_result = decrypt(custom_cipher, method='greedy')
custom_predicted_key = custom_result['key']
custom_decrypted = custom_result['plaintext']

print(f"\nPredicted key:  {custom_predicted_key}")
print(f"Decrypted:      {custom_decrypted.strip()}")

custom_key_acc = key_accuracy(custom_predicted_key, custom_true_key)
custom_char_acc = character_accuracy(custom_decrypted.strip(), custom_plaintext.strip())
print(f"\nKey accuracy:       {custom_key_acc:.2f}%")
print(f"Character accuracy: {custom_char_acc:.2f}%")

print("\n" + "="*80)
print("Interactive demo complete!")
print("="*80)
```
